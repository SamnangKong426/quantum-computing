```python
import ipywidgets as widgets
from IPython.display import display
from qiskit import QuantumCircuit

qc = QuantumCircuit(2)

def add_h(_):
    qc.h(0)
    display(qc.draw('mpl'))

button = widgets.Button(description="Add H gate on q0")
button.on_click(add_h)

display(button)
```
